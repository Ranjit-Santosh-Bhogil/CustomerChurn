Logging is important aspect of any application it is used to track the events,errors and operational
information.Pythons build in Logging module offers flexible framework for emitting log message from python programs 

```python
import logging
#Configure simple logging settings
logging.basicConfig(level=logging.DEBUG)
```

```python
#logging messages of different severity levels
logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
```
