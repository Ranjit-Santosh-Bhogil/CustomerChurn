```python
import sqlite3
```

SQLite is a self-contained,light weight,serverless database system

```python
connection = sqlite3.connect("example.db")
connection
```

```python
cursor = connection.cursor()

#Create a table
cursor.execute('''Create table if not exists emp(
    id Integer Primary Key,
    name text,
    salary Integer
)

''')

connection.commit()
```

```python
#Inserting data into the table
cursor.execute('''Insert into emp(id,name,salary) values

(2,'Jane',2000),
(3,'Bob',3000),
(4,'Alice',4000)
''') 

connection.commit()
```

```python
#Accessing the data from the table
cursor.execute('''
Select * from emp
''')
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
#Updating the data in the table
cursor.execute('''UPDATE emp SET salary = 1500 WHERE id = 1''')
connection.commit()
```

```python
#Getting the updated data from the table
cursor.execute("select * from emp")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
#Deleting data from the table
cursor.execute('''DELETE FROM emp WHERE id = 2''')
connection.commit()
```

```python
#Inserting data in bulk into the table
data = [(5, 'Charlie', 5000), (6, 'David', 6000), (7, 'Eve', 7000)]
cursor.executemany('''INSERT INTO emp(id, name, salary) VALUES (?, ?, ?)''', data)
connection.commit()
```

```python
#Getting complete data from the table after bulk insertion
cursor.execute("select * from emp")
rows = cursor.fetchall()
for row in rows:    
    print(row)
```

```python
connection.close()
#Now we cant access the data from the table as the connection is closed. If we try to access it, it will give an error.
```
